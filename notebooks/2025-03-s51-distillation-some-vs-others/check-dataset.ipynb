```python
from xaikd import datasets
import numpy as np
```

```python
dataset = datasets.construct("cifar100-people-vs-others")
```

```python
ds = datasets.subsample_dataset(
    dataset.create_subset(train_split=True),
    ratio=0.03125,
    seed=1
)
```

```python
arr_ys = []
for _, y in datasets.build_dataloader(ds, shuffle=False):
    arr_ys.extend(y.numpy().tolist())
    
arr_ys = np.array(arr_ys)
```

```python
np.bincount(arr_ys)
```
