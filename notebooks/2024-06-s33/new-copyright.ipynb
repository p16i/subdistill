```python
from xaikd import datasets

from xaikd.utils import spurious_feature_generator

from torchvision import transforms as T

from matplotlib import pyplot as plt
```

```python
actual_transforms = T.Compose([
    T.Resize(256),
    T.CenterCrop(224)
])
```

```python
img = datasets.imagenet.default_loader("../../datasets/imagenet/train/n01440764/n01440764_10026.JPEG")
```

```python
img
```

```python
alter = spurious_feature_generator.imagenet_copyright(img, seed=1)

alter
```

```python
actual_transforms(alter)
```

```python
plt.figure(figsize=(5*4, 5*2))
for i in range(4):
    plt.subplot(2, 4, i+1)
    obj = spurious_feature_generator.imagenet_copyright(img, seed=i)
    
    plt.imshow(obj)
    plt.subplot(2, 4, 4 + i+1)
    plt.imshow(actual_transforms(obj))
```
