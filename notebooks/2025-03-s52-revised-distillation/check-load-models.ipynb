```python
from xaikd import utils, models, datasets, metrics
```

```python
model = utils.modules.load_model_from_wandb_artifact(
    models.get_untrained_model("student-lenetxl", num_classes=1),
    run_path="p16i/kitchen-sink/5rf0u2rk",
    model_key="student",
    wandb_artifact_suffix="best",
)
```

```python
dataset = datasets.construct("cifar100-people-vs-others")
```

```python
dl_test = datasets.build_dataloader(
    dataset.create_subset(train_split=False),
    shuffle=False
)
```

```python
metric = metrics.MetricAUROCBinaryCrossEntropy()
```

```python
metric(model=model, dataloader=dl_test, device="cpu")
```
