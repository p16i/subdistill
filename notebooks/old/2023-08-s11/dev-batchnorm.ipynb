```python
from torch import nn

import torch
```

```python
x1= torch.randn((10, 2, 5, 5)) + 2
x2 = torch.randn((10, 2, 5, 5)) + 5
```

```python
bn1 = nn.BatchNorm2d(num_features=2)
bn2 = nn.BatchNorm2d(num_features=2, track_running_stats=False)
```

```python


for x in [x1, x2]:
    bn1(x)
    
    print(bn1.running_mean, bn1.running_var)
```

```python
for x in [x1, x2]:
    bn2(x)
    
    print(bn2.running_mean, bn2.running_var)
```

```python
bn2.train()
```

```python
for x in [x1, x2]:
    bn2(x)
    
    print(bn2.running_mean, bn2.running_var)
```
