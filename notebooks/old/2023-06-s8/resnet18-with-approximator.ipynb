```python
def get_approximator(layer_ix, out_dim):
    model = models._resnet18_cifar(100)
    model.inplanes = getattr(model, f"layer{layer_ix}")[0].conv1.weight.shape[1]
    
    return model._make_layer(
        resnet.BasicBlock,
        out_dim,
        1,
        stride=2, dilate=False
    )
    
get_approximator(3, 256)
```

```python
def ano():
    
    model = models.get_model("cifar100-resnet18-p1")
    model.layer4 = get_approximator(4, out_dim=512)
    
    model(inp)
ano()
```
