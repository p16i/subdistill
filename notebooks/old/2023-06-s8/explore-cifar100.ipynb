```python
from torchvision.datasets import CIFAR100

from torch.utils.data import DataLoader

from torchvision import transforms

import numpy as np 
```

```python
ds = CIFAR100(root="./datasets", train=True, download=True, transform=transforms.ToTensor())
```

```python
dl = DataLoader(ds)
```

```python
arr_y = []

for _, y in dl:
    arr_y.extend(y.tolist())
```

```python
np.bincount(arr_y)
```
