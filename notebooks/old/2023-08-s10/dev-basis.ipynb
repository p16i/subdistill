```python
import numpy as np
```

```python
np.random.seed(1)
x = np.random.randn(10, 3)
x[:, 2] = 10*x[:, 2]
x[:, 1] = 5*x[:, 1]

x
```

```python
def act_recon(a):
    
    norm = np.linalg.norm(x, axis=1,keepdims=True)
    
    criteria = norm ** 2 - 2 * a + a**2
    
    criteria = np.mean(criteria, axis=0)
    
    return np.argsort(-criteria)

act_recon(x)
```

```python
def act_recon_greedy(a):
    
    indices = []
    
    d = a.shape[1]
    
    while len(indices) < a.shape[1]:
        
        dimensions = list(set(range(a.shape[1])).difference(indices))
        norms = np.linalg.norm(a[:, dimensions], axis=1)
        
        stats = []
        print("--------")
        
        U = np.eye(d)[:, indices]
        
        a_c = a @ (np.eye(d) - U@U.T)
        for i in dimensions:
            u = np.zeros(d)
            u[i] = 1
            print(u)
            print(a.dot(u), a.dot(u).shape, u.shape)
            norm = np.linalg.norm(a_c - a.dot(u)[:, None]*u)
            print(norm)
            stat = np.mean(norm)
            print(f"[i={i}]: stat={stat}")
            stats.append(stat)
            
            
            
        selected_i = dimensions[np.argmin(stats)]
        print(f"picked={selected_i}")
        indices.append(selected_i)
        
    return indices
act_recon_greedy(x)
```
