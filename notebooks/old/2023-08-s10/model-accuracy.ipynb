```python
from xaikd import models, datasets
from xaikd.utils import metrics

import numpy as np

from tqdm import tqdm
```

```python
superclasses = """
cifar100-aquatic_mammals
cifar100-fish
cifar100-flowers
cifar100-food_containers
cifar100-fruit_and_vegetables
cifar100-household_electrical_devices
cifar100-household_furniture
cifar100-insects
cifar100-large_carnivores
cifar100-large_man_made_outdoor_things
cifar100-large_natural_outdoor_scenes
cifar100-large_omnivores_and_herbivores
cifar100-medium_mammals
cifar100-non_insect_invertebrates
cifar100-people
cifar100-reptiles
cifar100-small_mammals
cifar100-trees
cifar100-vehicles_1
cifar100-vehicles_2
""".strip().split("\n")
```

```python
superclasses
```

```python
def compute_acc(model_name, superclasses):
    stats = []
    model = models.get_model(model_name)
    
    for dataset_name in tqdm(superclasses):
        
        dataset = datasets.construct(dataset_name)
        
        val_ld = datasets.build_dataloader(
            dataset.create_subset(train_split=False),
            shuffle=False
        )
        acc = metrics.accuracy_with_subclasses(
            model=model,
            dl=val_ld,
            considered_classes=dataset.selected_classes,
            transform_target=dataset.transform_target,
            device="cpu"
        )
        print(f">>>>> {dataset_name}: acc={acc:.4f}")
        stats.append(acc)
        
    print(f"=== mean(acc)={np.mean(stats):.4f} ({len(superclasses)} classes)")

compute_acc("cifar100-resnet18-p1", superclasses)
```

```python
compute_acc("cifar100-resnet18-p2", superclasses)
```

```python
compute_acc("cifar100-resnet18-p3", superclasses)
```

```python
def compute_acc_cifar100(model_name):
    dataset_name = "cifar100"
    model = models.get_model(model_name)
    
    dataset = datasets.construct(dataset_name)
    
    
    for train_split in tqdm([True, False]):
        
        dl = datasets.build_dataloader(
            dataset.create_subset(train_split=train_split),
            shuffle=False
        )

        acc = metrics.accuracy(model, dl, num_classes=100, device="cpu")
    
        print(f"> {model_name}: acc[{dataset_name}(train_split={train_split})]={acc:.4f}")
    
compute_acc_cifar100("cifar100-resnet18-p1")
```

```python
compute_acc_cifar100("cifar100-resnet18-p2")
```

```python
compute_acc_cifar100("cifar100-resnet18-p3")
```

```python
compute_acc_cifar100("cifar100-vgg11-p1")
```

```python
compute_acc_cifar100("cifar100-resnet50-p1")
```
