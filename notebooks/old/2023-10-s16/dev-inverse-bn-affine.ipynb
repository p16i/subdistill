```python
import torch
from torch import nn

import numpy 
```

```python
torch.manual_seed(1)

x1 = torch.randn(16, 5, 3, 3) + 5
x2 = torch.randn(16, 5, 3, 3)
```

```python
bn = nn.BatchNorm2d(5)
model = nn.Sequential(
    bn
)
```

```python
for i in range(10):
    _ = model(x1)
```

```python
model.eval()
```

```python
out1 = model(x2)
```

```python
actual = (out1 - bn.bias.view(1, -1, 1, 1)) / (bn.weight.view(1, -1, 1, 1))
```

```python
expected = (x2 - bn.running_mean.view(1, -1, 1, 1)) / ( (bn.running_var.view(1, -1, 1, 1) + bn.eps) ** 0.5 )
```

```python
numpy.testing.assert_allclose(
    actual.detach(),
    expected.detach(),
    atol=1e-6
)
```
