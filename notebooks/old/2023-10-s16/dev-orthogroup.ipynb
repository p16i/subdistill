```python
from scipy.stats import ortho_group

import numpy as np 

from matplotlib import pyplot as plt
```

```python
k = 64
```

```python
U1 = ortho_group.rvs(k, random_state=np.random.default_rng(1))

U1[0, :5]
```

```python
U2 = ortho_group.rvs(k*2, random_state=np.random.default_rng(1))
```

```python
U1.shape
```

```python
U2.shape
```

```python
plt.imshow(U1@U2[:64, :64])
plt.colorbar()
```
