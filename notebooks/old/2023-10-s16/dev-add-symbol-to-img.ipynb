```python
from PIL import Image, ImageDraw, ImageFont

import itertools

from matplotlib import pyplot as plt
```

```python
def ano():
    arr_x = (0, 31-4)
    arr_y = (-3, 31-4-3)
    
    size = (32, 32)
    
    locations = list(itertools.product(arr_x, arr_y))
    ncols = len(locations)
    
    plt.figure(figsize=(1*ncols, 1))
    
    for i, loc in enumerate(locations):
        plt.subplot(1, ncols, i +1)
        
    
        im = Image.new("RGB", size, "white")
        d = ImageDraw.Draw(im)

        d.text(loc, "+", fill="red")
    
        plt.imshow(im)
        if i == 0:
            plt.ylabel(f"{size}", fontsize="small")
        plt.title(loc, fontsize="small")
        plt.xticks([])
        plt.yticks([])
ano()
```
