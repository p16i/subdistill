```python
from xaikd import datasets

from xaikd.showcases import cleverhans

import numpy as np

from matplotlib import pyplot as plt
```

```python
def ano():
    
    dataset = datasets.construct("cifar100-people")
    ds_train = datasets.subsample_dataset(dataset.create_subset(train_split=True), 0.05, seed=1)

    victim_class_ix = np.min(ds_train.dataset.targets)

    ds_contaminated = cleverhans.contaminate_dataset(
        ds_train,
        contamination_level=1.0,
        seed=3,
        victim_class_indices=[victim_class_ix]
    )

    
    indices_samples_of_victim_class = list(set(ds_contaminated.indices).intersection(
        np.argwhere(np.array(ds_contaminated.dataset.targets) == victim_class_ix).reshape(-1)
    ))
    
    selected_ix = indices_samples_of_victim_class[-5]
    
#     indices_samples_of_victim_class = [selected_ix]
    indices_samples_of_victim_class = indices_samples_of_victim_class
    nrows = len(indices_samples_of_victim_class)
    ncols = 2
    plt.figure(figsize=(ncols*1.8, nrows*1.8))
    
    for _i, idx in enumerate(indices_samples_of_victim_class):
        
        img = ds_contaminated.dataset.data[idx]
        clean_img = ds_train.dataset.data[idx]
        y = ds_contaminated.dataset.targets[idx]
        is_victim = not (img == clean_img).all()

        plt.subplot(nrows, ncols, _i*ncols +1)
        if _i == 0:
            plt.title("Clean" if is_victim else "")

        plt.imshow(clean_img); plt.xticks([]); plt.yticks([])
        plt.ylabel(_i)
        plt.subplot(nrows, ncols, _i*ncols + 2)
        if _i == 0:
            plt.title("Contaminated" if is_victim else "")
        plt.imshow(img); plt.xticks([]); plt.yticks([])
        

        
#     plt.savefig(OUTPUT_DIR / "contamination-example.png", dpi=150)

ano()
```
