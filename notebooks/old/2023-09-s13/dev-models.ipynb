```python
model = models.get_trained_model("cifar100-resnet18-wb15")
```

```python
from xaikd import datasets

from xaikd.utils import metrics
```

```python
dataset = datasets.construct("cifar100-people")
```

```python
ds_val = dataset.create_subset(train_split=False)
```

```python
from xaikd import utils
```

```python
utils.modify_last_layer_for_subclasses(model.fc, dataset.selected_classes)
```

```python
metrics.accuracy(
    model,
    datasets.build_dataloader(ds_val, shuffle=False), 5, "cpu"
)
```
