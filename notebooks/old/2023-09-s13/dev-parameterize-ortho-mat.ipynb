```python
import torch
from torch.nn.utils.parametrizations import orthogonal

from torch import nn
```

```python
lin = nn.Conv2d(20, 40, kernel_size=1)
torch.linalg.matrix_norm(lin.weight.squeeze(), ord=2)
```

```python
orth_linear = orthogonal(lin)
```

```python
Q = orth_linear.weight.squeeze()
```

```python
torch.dist(Q.T @ Q, torch.eye(20))
```

```python
torch.linalg.matrix_norm(Q, ord=2)
```
