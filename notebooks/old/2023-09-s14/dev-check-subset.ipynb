```python
from torch.utils.data import random_split
import torch
from torch.utils.data import TensorDataset, DataLoader, Subset
```

```python
ds = TensorDataset(torch.tensor(range(10)))
```

```python
ds_1, ds_2 = random_split(ds, [7, 3], generator=torch.Generator().manual_seed(1))
```

```python
for x in DataLoader(ds_1, batch_size=10, shuffle=False):
    print(x)
    
for x in DataLoader(ds_2, batch_size=10, shuffle=False):
    print(x)
```

```python
ss = Subset(ds, indices=range(5))

ss_1, ss_2 = random_split(ss, [0.7, 0.3], generator=torch.Generator().manual_seed(1))

for x in DataLoader(ss_1, batch_size=10, shuffle=False):
    print(x)
    
for x in DataLoader(ss_2, batch_size=10, shuffle=False):
    print(x)
```

```python
from xaikd import datasets, models
from xaikd.utils import metrics
from xaikd import utils
```

```python
dataset = datasets.construct("cifar100-people")
```

```python
model = models.get_trained_model("cifar100-resnet18-v1")
```

```python
utils.modify_last_layer_for_subclasses(model, dataset.selected_classes)
```

```python
train_ds = dataset.create_subset(train_split=True)
```

```python
metrics.accuracy(model, datasets.build_dataloader(train_ds, shuffle=False), num_classes=5, device="cpu")
```

```python
subsmapled_train_ds = datasets.subsample_dataset(train_ds, ratio=1, seed=1)
```

```python
metrics.accuracy(model, datasets.build_dataloader(subsmapled_train_ds, shuffle=False), num_classes=5, device="cpu")
```
