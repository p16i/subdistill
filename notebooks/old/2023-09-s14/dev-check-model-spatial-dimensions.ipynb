```python
import torch

from xaikd import models, utils
```

```python
def ano(model, layers=["layer1", "layer2", "layer3", "layer4"]):
    x = torch.randn((10, 3, 32, 32))
    
    
    with torch.no_grad():
        (
            teacher_logits,
            teacher_arr_intermediate_feats,
        ) = utils.interceptor.forward_and_intercept_intermediate_layers(
            model,
            x,
            layers=layers
        )
    
    for lix, (layer, feat) in enumerate(zip(layers, teacher_arr_intermediate_feats)):
        print(f"[layer={layer}]: feat.shape={feat.shape}")
        
ano(models.get_trained_model("cifar100-resnet18-v1"))
```

```python
ano(models.get_trained_model("cifar100-resnet50-v1"))
```

```python
ano(models.get_trained_model("cifar100-vgg11-v1"), layers=[
    "features.10",
    "features.15",
    "features.20"
])
```

```python
ano(models.get_untrained_model("vgg8", num_classes=100), layers=[
    "features.8",
    "features.11",
    "features.14",
#     "features.20"
])
```
