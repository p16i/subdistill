```python
from xaikd.distillators import ModelWrapper
from xaikd import datasets, models

from xaikd.utils import metrics

from xaikd import attributors
from xaikd.showcases import cleverhans

from matplotlib import pyplot as plt
import numpy as np 

from torch import nn
from torch.nn import functional as F
```

```python
dataset = datasets.construct("cifar100-people")
```

```python
dl_clean = datasets.build_dataloader(
    cleverhans.contaminate_dataset(
        dataset=datasets.subsample_dataset(
            dataset=dataset.create_subset(train_split=False), ratio=1.0, seed=1
        ),
        contamination_level=0.0,
        seed=1,
        victim_class_indices=dataset.selected_classes,
    ),
    shuffle=False,
    batch_size=64
)
```

```python
dl_contaminated = datasets.build_dataloader(
    cleverhans.contaminate_dataset(
        dataset=datasets.subsample_dataset(
            dataset=dataset.create_subset(train_split=False), ratio=1.0, seed=1
        ),
        contamination_level=0.75,
        seed=1,
        victim_class_indices=dataset.selected_classes,
    ),
    shuffle=False,
    batch_size=64
)
```

```python
def ano(run_id):

    artifact = run.use_artifact(f'p16i/kitchen-sink/model-{run_id}:v0', type='model')
    artifact_dir = artifact.download()
    
    checkpoint = torch.load(f"{artifact_dir}/model.ckpt")
    approximator = checkpoint["approximator"]
    adapter = checkpoint["adapter"]
    
    in_chs, out_chs, _, _ = adapter.mat_decoder.shape
    
    adapter_convereted = nn.Conv2d(in_channels=in_chs, out_channels=out_chs, kernel_size=1)
    adapter_convereted.weight = nn.Parameter(adapter.mat_decoder)
    adapter_convereted.bias = nn.Parameter(adapter.mean.squeeze())
    
    model = models.get_model("cifar100-resnet18-wb15")
#     model = models.get_model("cifar100-resnet18-p1")

    setattr(
        model, 
        "layer3", 
        nn.Sequential(
            approximator,
            adapter_convereted
        )
    )

    model.eval();
    
    print(f"[runid={run_id}]")
    
    
    
    for label, dl in [
        ("clean", dl_clean),
        ("contamined", dl_contaminated)
    ]:
            
        acc, xent = metrics.accuracy_with_subclasses(
            model, dl,
            considered_classes=dataset.selected_classes,
            transform_target=dataset.transform_target,
            device="cpu",
            verbose=False
        )
        
        print(f"[{label}]: acc={acc:.4f} | xent={xent:.4f}")
        
        x, y = next(iter(dl))
        
        x = x[:20]
        y = y[:20]    

        ncols = x.shape[0]
        nrows = 2


        logit_mod = attributors.OneClassEvidence(dataset=dataset)


        with attributors.make_attributor_for(model, dataset.input_statistics) as attributor:
                logits, heatmaps = attributor.forward(x, lambda logits: logit_mod(logits, y))

        heatmaps = heatmaps.sum(axis=1).numpy()

        plt.figure(figsize=(ncols*1.3, nrows*1.3))

        
        transformed_y = dataset.transform_target(y)
        transformed_logit = logits.detach().numpy()[:, dataset.selected_classes]

        for i in range(ncols):
            plt.subplot(nrows, ncols, i+1)
            img = x[i].permute(1, 2, 0).numpy()
            img = (img  * dataset.input_statistics[1] + dataset.input_statistics[0])
            img = np.clip(img, a_min=0, a_max=1)

            target = transformed_y[i]
            pred = np.argmax(transformed_logit[i])
            heatmap = heatmaps[i]        
            suffix = "C" if target==pred else "F"

            plt.title(f"{target} ➞ {pred}", fontsize="small")
            plt.imshow(img); plt.xticks([]); plt.yticks([])

            plt.subplot(nrows, ncols, ncols + i+1)
            plt.imshow(heatmap)
            plt.xticks([]); plt.yticks([])
            plt.xlabel(f"{heatmap.sum():02.2f} ({logits[i][target]:02.2f})")

ano("ap7vdo96") # identity-compr1.0, alpha=0
```
