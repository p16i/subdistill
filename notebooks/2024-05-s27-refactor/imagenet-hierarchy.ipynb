```python
r = requests.get("https://gist.githubusercontent.com/mbostock/535395e279f5b83d732ea6e36932a9eb/raw/62863328f4841fce9452bc7e2f7b5e02b40f8f3d/mobilenet.json")
data = r.json()
```

```python
def tranverse(root, cutoff=10, cache={}):
    # is paranet of node?
    root_name = root["name"]
    cache[root_name] = True

    is_parent_of_node = True
    for child in root["children"]:
        if "children" in child and len(child["children"]) > 0:
            is_parent_of_node = False
            if child["name"] not in cache:
                tranverse(child, cutoff=cutoff, cache=cache)

    count_children = len(root["children"]) 
    if is_parent_of_node \
        and (7 <= count_children <= 10) \
        and root:
        print(f"==== {root_name} (total={count_children}) ====")
        for cix, child in enumerate(root["children"]):
            assert "children" not in child
            child_name = child["name"]
            child_nsid = child["id"]
            print(f"> [ix={cix:2d}] {child_name} ({child_nsid})")

tranverse(data, cutoff=5)
```
