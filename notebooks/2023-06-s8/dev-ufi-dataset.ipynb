```python
import os

from torchvision.datasets import VisionDataset

from pathlib import Path

from glob import glob

from PIL import Image

from matplotlib import pyplot as plt
```

```python
class UFI(VisionDataset):
    def __init__(self, root, train, *args, **kwargs):
        super().__init__(root, *args, **kwargs) 
        
        num_classes = 605
        
        slug = "train" if train else  "test"
        
        dirname = Path(root) / slug / "s*" / "*.pgm"
        print(dirname)
        self.filepaths = glob(str(dirname))
        
    
    def __getitem__(self, idx):
        filepath = self.filepaths[idx]
        img = Image.open(filepath)
        
        slug = os.path.split(
            os.path.dirname(filepath)
        )[-1]
        
        slug = int(slug.replace("s", ""))
        
        return img, slug
        
    def __len__(self):    
        return len(self.filepaths)
ds = UFI(root="/home/pat/Downloads/ufi-cropped", train=True)
```
