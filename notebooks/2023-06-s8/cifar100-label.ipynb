```python
def ano():
    
    
    with open("../../datasets/cifar100/cifar-100-python/meta", "rb") as fh:
        meta = pickle.load(fh)
        
    with open("../../datasets/cifar100/cifar-100-python/train", "rb") as fh:
        data = pickle.load(fh, encoding='latin1')
        fine_labels = data["fine_labels"]
        coarse_labels = data["coarse_labels"]
    arr_rows = []
    for fine_label, coarse_label in set(zip(fine_labels, coarse_labels)):
        arr_rows.append(dict(
            coarse_label_name=meta["coarse_label_names"][coarse_label].replace("-", "_"),
            fine_label_name=meta["fine_label_names"][fine_label].replace("-", "_"),
            fine_label=fine_label,
            coarse_label=coarse_label,
        ))
    return pd.DataFrame(arr_rows)
df = ano()
```
