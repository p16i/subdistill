```python
from xaikd import models, utils
import torch
```

```python
def pretty_print2(model, arr_layers, input_size=(1, 3, 224, 224)):    
    
    x = torch.randn(*input_size)
    _, arr_outputs = utils.interceptor.forward_and_intercept_intermediate_layers(
        model, x,
        layers=arr_layers,
        detach_output=False
    )
    
    mapping = dict()

    for lix, output in enumerate(arr_outputs):
        mapping[arr_layers[lix]] = output.shape
    return mapping
pretty_print2(models.get_trained_model("imagenet-vitb-tv"), ["encoder.layers.8", "encoder.layers.11"])
```

```python
pretty_print2(models.get_untrained_model("vitstudent-48", num_classes=10), ["encoder.layers.2", "encoder.layers.3"])
```

```python
pretty_print2(models.get_untrained_model("vitstudent-72", num_classes=10), ["encoder.layers.2", "encoder.layers.3"])
```

```python
def pretty_print(model, input_size):    
    num_features = len(model.features)
    print(f"----- ({num_features} features) ------")
    
    x = torch.randn(*input_size)
    _, arr_outputs = utils.interceptor.forward_and_intercept_intermediate_layers(
        model, x,
        layers=[f"features.{ix}" for ix in range(num_features)],
        detach_output=False
    )

    for lix, output in enumerate(arr_outputs):
        print(f"features.{lix}: {output.shape}")

pretty_print(models.get_untrained_model("student-mobilenets", num_classes=10), input_size=(7, 3, 32, 32))
```

```python
pretty_print2(models.get_trained_model("cifar100-resnet18-v1"), input_size=(7, 3, 32, 32), arr_layers=[
    "layer1", "layer2", "layer3", "layer4"])
```

```python
pretty_print2(models.get_untrained_model("student-32-24-16-8", num_classes=10), input_size=(7, 3, 32, 32), arr_layers=["layer3", "layer4"])
```

```python
pretty_print(models.get_untrained_model("student-mobilenetxxs-cifar", num_classes=10), input_size=(7, 3, 32, 32))
```
