```python
import numpy as np 
```

```python
def first_direction(a, c):
    
    a = a.reshape(1, 2)
    c = c.reshape(1, 2)
    
    mat = a.T @ c + c.T @ a 
    print(mat)
    eigvals, eigvecs = np.linalg.eigh(mat)
    print(eigvals)
    
    return eigvecs[:, -1]
```

```python
A = np.array([50, 20])
R = np.array([5, 2])
C = R /  A 

C
```

```python
first_direction(A, C)
```

```python
first_direction(A / np.linalg.norm(A), C / np.linalg.norm(C))
```

```python
An = A / np.linalg.norm(A)
Rn = R / np.linalg.norm(R)

Cn = Rn / An
An, Cn
```

```python
first_direction(An, Cn)
```

```python
Cnb = R / An

Cnb
```

```python
first_direction(An, Cnb)
```
