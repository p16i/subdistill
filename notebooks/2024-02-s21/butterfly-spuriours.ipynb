```python
from xaikd import datasets

from torchvision import transforms as T

from matplotlib import pyplot as plt
```

```python
rc_transform = T.Compose(
    [T.Resize(256), T.CenterCrop(224)]
)
```

```python
ds_butterfly = datasets.construct("imagenet-butterfly")
ds_clean_val = ds_butterfly.create_subset(train_split=False)
ds_clean_val.transform = rc_transform

ds_clean_val[1][0]
```

```python
plt.imshow(ds_clean_val[1][0]); plt.xticks([]); plt.yticks([])
plt.savefig("./figures-tac3/imagenet-butterfly/clean.jpg")
```

```python
ds_butterfly_spurious = datasets.construct("imagenet-butterfly-spurious0.5")
ds_spurious_val = ds_butterfly_spurious.create_subset(train_split=False)
ds_spurious_val.transform = rc_transform

ds_spurious_val[1][0]
```
