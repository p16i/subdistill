```python
import numpy as np

from matplotlib import pyplot as plt

import pandas as pd

import json
```

```python
def load(basis_name, layer):
    with open(
        f"../../artifacts/2024-09-ipam/experiments/accuracy-basis/cifar100/cifar100-resnet18-v1/{layer}/{basis_name}--uncentered/accuracy.json", 
        "r"
    ) as fh:
        return json.load(fh)
load("pcalookahead", "layer1")
```

```python
def viz():

    arr_layers = ["layer1", "layer2", "layer3", "layer4"]
    
    arr_attrs = ["accuracies", "losses"]
    arr_bases = [
        "pca", 
        "pcalookahead",
        "prca-sortabs"
    ]

    ncols = len(arr_layers)
    nrows = len(arr_attrs)
    plt.figure(figsize=(3*len(arr_layers), 3*nrows))
    plt.suptitle("ResNet18: CIFAR100")
    
    for cix, layer in enumerate(arr_layers):
        for bix, basis in enumerate(arr_bases):
            for aix, attr in enumerate(arr_attrs):
                plt.subplot(nrows, ncols, cix + 1 + aix*ncols)
                

                data = load(basis_name=basis, layer=layer)
                d = data["arr_ks"][-1]
                plt.plot(data["arr_ks"], data[attr], label=basis)
                plt.axhline(data[attr][-1], ls="--", lw=1, color="k")

                if aix == 1:
                    plt.xlabel(f"Subspace Dimensions $k$\n(d={d})")
                
                if cix == 0:
                    plt.ylabel(f"val-{attr}")

                if aix == 0:
                    plt.title(layer)
                    plt.xticks([])
                if aix == 0 and cix == 0:
                    plt.legend()
viz()
```
